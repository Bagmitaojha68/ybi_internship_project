```python
from pathlib import Path
import pandas as pd

# FIXED PATH
root = Path(r"C:\Users\Bagmita\OneDrive\Desktop\Project")
files = list(root.rglob("day38_relationship_analyzed_dataset.csv"))

print(f"Searching in: {root}")

if files:
    df = pd.read_csv(files[0])
    print("Dataset loaded:", files[0])
    print("Shape:", df.shape)
    display(df.head())
else:
    print("Dataset not found. Day 38 CSV file check kara.")
    for f in root.rglob("*.csv"):
        print(f.name)
```

```python
features = [
    "study_hours",
    "attendance",
    "previous_score",
    "assignments_completed"
]

available = [col for col in features if col in df.columns]

print("Selected features:", available)
```

```python
variability = df[available].var().sort_values(ascending=False)

importance_report = variability.reset_index()
importance_report.columns = ["Feature", "Variance"]

display(importance_report)
```

```python
importance_report.plot(
    x="Feature",
    y="Variance",
    kind="bar",
    legend=False,
    figsize=(8, 5)
)

plt.title("Feature Variability Analysis")
plt.ylabel("Variance")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
importance_report.to_csv(
    "day39_feature_variability_report.csv",
    index=False
)

print("Feature analysis report saved!")
```
