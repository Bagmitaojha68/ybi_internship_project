```python
from pathlib import Path
import pandas as pd

# 1. Sahi bada dataset khoja
root = Path(r"C:\Users\Bagmita\OneDrive\Desktop\Project")
files = list(root.rglob("day36_model_ready_dataset.csv"))
if not files:
    files = list(root.rglob("*model_ready*.csv"))

print(f"Using file: {files[0]}")
df = pd.read_csv(files[0])
print("Original Shape:", df.shape)
print("Columns:", df.columns.tolist())

# 2. Target column create kara - YBI project re
# Jodi previous_score ru performance banaba
# Example logic:
if 'predicted_performance' not in df.columns:
    # Jodi previous_score 60 ru besi ta Good, nahale Average
    # Tuma original data re final_score thile seta use kara
    if 'previous_score' in df.columns:
        df['predicted_performance'] = df['previous_score'].apply(lambda x: 'Good' if x >= 60 else 'Average' if x >= 40 else 'Poor')
        print("\n'predicted_performance' column create hela!")

# 3. Ebe check kara
target = "predicted_performance"
if target in df.columns:
    print(f"\nTarget column found: {target}")
    print(df[target].value_counts(dropna=False))
    display(df.head())
else:
    print("Target column not found.")
```

```python
if target in df.columns:
    print("Target data type:", df[target].dtype)
    print("Unique target values:", df[target].nunique())
    print("Missing target values:", df[target].isnull().sum())

    print("\nImportant:")
    print("Confirm whether this column contains actual outcomes")
    print("or predictions generated by an earlier model.")
```

```python
feature_report = pd.DataFrame({
    "Feature": available_features,
    "Data Type": [df[col].dtype for col in available_features],
    "Missing Values": [
        df[col].isnull().sum() for col in available_features
    ]
})

feature_report.to_csv(
    "day40_feature_preparation_report.csv",
    index=False
)

display(feature_report)
print("Feature report saved!")
```
