```python
from pathlib import Path
import pandas as pd

# Tuma laptop ra sahi path
root = Path(r"C:\Users\Bagmita\OneDrive\Desktop\Project")

# Sabu CSV khojiba
files = list(root.rglob("day37_analyzed_student_features.csv"))

print(f"Khojuthiba jaga: {root}")

if files:
    df = pd.read_csv(files[0])
    print("Dataset loaded:", files[0])
    print("Shape:", df.shape)
    display(df.head())
else:
    print("Dataset not found. Day 37 CSV file check kara.")
    print("Project re thiba sabu csv:")
    for f in root.rglob("*.csv"):
        print(f.name)
```

```python
features = [
    "study_hours",
    "attendance",
    "previous_score",
    "assignments_completed"
]

available = [col for col in features if col in df.columns]

print("Features selected:", available)
```

```python
corr = df[available].corr()

display(corr.round(2))
```

```python
plt.figure(figsize=(8, 6))

plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")

plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.columns)), corr.columns)

for i in range(len(corr)):
    for j in range(len(corr)):
        plt.text(j, i, f"{corr.iloc[i, j]:.2f}", ha="center", va="center")

plt.title("Student Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```

```python
corr.to_csv("day38_feature_correlation_matrix.csv")

print("Correlation matrix saved successfully!")
```
