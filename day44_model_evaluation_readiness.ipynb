```python
from pathlib import Path
import pandas as pd

# FIX - Permanent
root = Path.cwd()
print(f"Current folder: {root}")

# Find Day 41 file anywhere
files = list(root.rglob("day41_validated_training_features.csv"))
if not files:
    files = list(root.glob("*.csv"))

if files:
    # Prefer validated file
    target = None
    for f in files:
        if "day41" in f.name and "validated" in f.name:
            target = f
            break
    if not target:
        target = files[0]
    
    df = pd.read_csv(target)
    print(f"Dataset loaded successfully!")
    print(f"Shape: {df.shape}")
    display(df.head())
else:
    print("File not found - check Day 41 CSV in folder")
```

```python
target = "actual_performance"

if target in df.columns:
    print("Actual target column is available.")
else:
    print("Actual target column is missing.")
    print("Model evaluation cannot be performed reliably.")
```

```python
readiness_report = pd.DataFrame({
    "Check": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows",
        "Actual Target Available"
    ],
    "Result": [
        len(df),
        len(df.columns),
        int(df.isnull().sum().sum()),
        int(df.duplicated().sum()),
        target in df.columns
    ]
})

display(readiness_report)
```

```python
output_dir = Path.cwd()

readiness_report.to_csv(
    output_dir / "day44_model_evaluation_readiness_report.csv",
    index=False
)

print("Day 44 report saved successfully!")
```
