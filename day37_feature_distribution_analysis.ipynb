```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

print("Student Performance Prediction System")
print("Day 37: Feature Distribution Analysis")
```

```python
root = Path(r"C:\Users\Bagmita\OneDrive\Desktop\Project")
files = list(root.rglob("day36_model_ready_dataset.csv"))

if files:
    df = pd.read_csv(files[0])
    print("Dataset loaded:", files[0])
    print("Shape:", df.shape)
    display(df.head())
else:
    print("Dataset not found. Day 36 CSV file check kara.")
    print(f"Checked in: {root}")
```

```python
features = [
    "study_hours",
    "attendance",
    "previous_score",
    "assignments_completed"
]

available = [col for col in features if col in df.columns]

print("Available features:", available)
display(df[available].describe().T)
```

```python
df[available].hist(figsize=(10, 7), bins=10)
plt.suptitle("Student Feature Distributions")
plt.tight_layout()
plt.show()
```

```python
skew_report = df[available].skew().reset_index()
skew_report.columns = ["Feature", "Skewness"]

display(skew_report)
```

```python
df.to_csv("day37_analyzed_student_features.csv", index=False)

print("Dataset saved successfully!")
print("Records:", len(df))
print("Features:", len(df.columns))
```
