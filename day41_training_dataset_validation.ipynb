```python
from pathlib import Path
import pandas as pd

# 1. LOAD - Sabu ekathare
root = Path(r"C:\Users\Bagmita\OneDrive\Desktop\Project")
df = pd.read_csv(root / "day40_prepared_student_features.csv")
X = df.copy()  # Eita important - X define heba

print("Dataset loaded:", X.shape)
print("Columns:", list(X.columns))

# 2. VALIDATION - YBI template requirement
required = ["study_hours", "attendance", "previous_score", "assignments_completed"]

# Jodi kichhi column nahi ta check kara
missing = [col for col in required if col not in X.columns]
print("Missing columns:", missing)
print("Available columns:", list(X.columns))

# 3. Jodi missing nahi ta age jiba
if not missing:
    from sklearn.model_selection import train_test_split
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.metrics import accuracy_score, classification_report
    import joblib

    # Features & Target - Tuma nua dataset re extra columns achi
    feature_cols = ["study_hours", "attendance", "previous_score", "assignments_completed", "internet_hours", "sleep_hours"]
    # Jodi extra column nahi thiba ta required use kara
    feature_cols = [c for c in feature_cols if c in X.columns]
    
    X_features = X[feature_cols]
    y = X["predicted_performance"]

    X_train, X_test, y_train, y_test = train_test_split(X_features, y, test_size=0.2, random_state=42)

    model = RandomForestClassifier(n_estimators=100, random_state=42)
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    acc = accuracy_score(y_test, y_pred)

    print(f"\n✅ Model Trained! Accuracy: {acc*100:.2f}%")
    print(classification_report(y_test, y_pred))

    # Save
    joblib.dump(model, root / "day41_trained_model.pkl")
    print("Model saved!")
else:
    print("File thik nahi, upara Step 1 check kara")
```

```python
validation_report = pd.DataFrame({
    "Feature": required,
    "Data Type": [X[col].dtype for col in required],
    "Missing Values": [X[col].isnull().sum() for col in required],
    "Unique Values": [X[col].nunique() for col in required]
})

display(validation_report)

validation_report.to_csv(
    "day41_training_validation_report.csv",
    index=False
)

print("Validation report saved!")
```
