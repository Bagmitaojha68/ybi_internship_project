```python
from pathlib import Path
import pandas as pd

root = Path(r"C:\Users\Bagmita\OneDrive\Desktop\Project")

# 1. Sabu csv dekha
print("Sabuthu badi file kana?")
for f in root.rglob("*.csv"):
    try:
        r = pd.read_csv(f).shape[0]
        print(f.name, "->", r, "rows")
    except:
        pass

# 2. Thik dataset load kara - Day 34 wala final features
files = list(root.rglob("day34_final_student_features.csv"))
if not files:
    files = list(root.rglob("day35_final_dataset*.csv"))
if not files:
    files = list(root.rglob("day35_final_dataset*.csv"))

print("\nLoading:", files[0])
df = pd.read_csv(files[0])
print(df.shape)
print(df.columns.tolist())
display(df.head())
```

```python
required = [
    "study_hours",
    "attendance",
    "previous_score",
    "assignments_completed"
]

missing = [col for col in required if col not in df.columns]

if not missing:
    print("All required features are available!")
else:
    print("Missing features:", missing)
```

```python
if all(col in df.columns for col in required):
    display(df[required].describe().T)
else:
    print("Required features check.")
```

```python
report = pd.DataFrame({
    "Check": [
        "Dataset Loaded",
        "Required Features Available",
        "Missing Values",
        "Duplicate Rows"
    ],
    "Result": [
        not df.empty,
        all(col in df.columns for col in required),
        int(df.isnull().sum().sum()),
        int(df.duplicated().sum())
    ]
})

report.to_csv("day36_model_readiness_report.csv", index=False)
display(report)

print("Day 36 report saved successfully!")
```

```python
output = "day36_model_ready_dataset.csv"

df.to_csv(output, index=False)

print("Dataset saved:", output)
print("Final shape:", df.shape)
print("Day 36 completed!")
```
